Step 1: Imports

```python
import torch
import torch.nn as nn
import torch.optim as optim
import numpy as np
import matplotlib.pyplot as plt
```

Step 2: Quantum Utilities

```python
# Function to create a random density matrix of size dim x dim
def random_density_matrix(dim=2):
    # Random complex matrix
    A = np.random.randn(dim, dim) + 1j * np.random.randn(dim, dim)
    rho = A @ A.conj().T
    rho /= np.trace(rho)  # ensure unit trace
    return rho

# Function to simulate measurement outcomes (Pauli measurements simplified)
def simulate_measurements(rho, num_measurements=10):
    # Flatten real and imaginary parts as input features
    features = np.concatenate([np.real(rho).flatten(), np.imag(rho).flatten()])
    # Add small Gaussian noise to simulate measurements
    measurements = features + 0.05 * np.random.randn(len(features))
    return measurements

# Quantum fidelity between two density matrices
def fidelity(rho, sigma):
    # sqrtm method using torch.linalg
    sqrt_rho = torch.linalg.matrix_power(rho, 1)  # for small matrices, use identity
    fid = torch.trace(torch.linalg.sqrtm(sqrt_rho @ sigma @ sqrt_rho)) ** 2
    return fid.real

# Trace distance between two density matrices
def trace_distance(rho, sigma):
    diff = rho - sigma
    td = 0.5 * torch.linalg.norm(diff, ord='nuc')  # nuclear norm = sum of singular values
    return td
```

Step 3: Transformer Model

```python
class DensityMatrixTransformer(nn.Module):
    def __init__(self, input_dim, output_dim=4, nhead=2, num_layers=1):
        super().__init__()
        self.input_dim = input_dim
        self.output_dim = output_dim  # output size for lower triangular elements
        self.embedding = nn.Linear(input_dim, 16)  # embed input
        encoder_layer = nn.TransformerEncoderLayer(d_model=16, nhead=nhead)
        self.transformer = nn.TransformerEncoder(encoder_layer, num_layers=num_layers)
        self.fc_out = nn.Linear(16, output_dim)

    def forward(self, x):
        # x: batch_size x input_dim
        x = self.embedding(x).unsqueeze(1)  # add seq dimension for transformer
        x = self.transformer(x)
        x = x.squeeze(1)
        L_flat = self.fc_out(x)
        # Build lower triangular matrix L for 2x2 example
        L = torch.zeros(x.size(0), 2, 2, dtype=torch.cfloat)
        L[:,0,0] = L_flat[:,0]
        L[:,1,0] = L_flat[:,1]
        L[:,1,1] = L_flat[:,2]
        return L
```

Step 4: Cholesky Reconstruction

```python
def reconstruct_density(L):
    # Compute rho = L L†
    rho = torch.matmul(L, torch.conj(L.transpose(-2, -1)))

    # Compute trace for each matrix in the batch
    # torch.diagonal sums diagonal elements along last two dims
    tr = torch.diagonal(rho, dim1=-2, dim2=-1).sum(-1)

    # Normalize each matrix to have unit trace
    rho = rho / tr.view(-1, 1, 1)

    return rho
```

Step 5: Training Setup

```python
# Hyperparameters
num_samples = 500
input_dim = 8  # real + imaginary flattened 2x2 matrix
lr = 0.01
num_epochs = 50

# Generate dataset
X = []
Y = []
for _ in range(num_samples):
    rho = random_density_matrix(2)
    meas = simulate_measurements(rho)
    X.append(meas)
    Y.append(rho)

X = torch.tensor(X, dtype=torch.float32)
Y = torch.tensor(Y, dtype=torch.cfloat)

# Model, loss, optimizer
model = DensityMatrixTransformer(input_dim=input_dim)
optimizer = optim.Adam(model.parameters(), lr=lr)
loss_fn = nn.MSELoss()  # simple loss on L reconstruction
```

Step 6: Training Loop

```python
for epoch in range(num_epochs):
    model.train()
    optimizer.zero_grad()

    # Forward
    L_pred = model(X)
    rho_pred = reconstruct_density(L_pred)

    # Compute loss (MSE between predicted and true rho)
    loss = torch.mean(torch.abs(rho_pred - Y)**2)

    # Backprop
    loss.backward()
    optimizer.step()

    if (epoch+1) % 10 == 0:
        print(f"Epoch {epoch+1}/{num_epochs}, Loss: {loss.item():.6f}")

# After training, check some example
model.eval()
with torch.no_grad():
    L_pred = model(X[:5])
    rho_pred = reconstruct_density(L_pred)
    print("Predicted rho[0]:\n", rho_pred[0])
    print("True rho[0]:\n", Y[0])
```

Step 7: Compute Metrics

```python
def fidelity(rho, sigma):
    """
    Quantum fidelity F(rho, sigma) = (Tr(sqrt(sqrt(rho) * sigma * sqrt(rho))))^2
    Works for small 2x2 or 4x4 matrices using eigendecomposition.
    """
    # Ensure Hermitian
    rho = (rho + rho.conj().T) / 2
    sigma = (sigma + sigma.conj().T) / 2

    # Eigen decomposition of rho
    eigvals, eigvecs = torch.linalg.eigh(rho)

    # Convert eigenvalues to complex for matmul
    sqrt_D = torch.diag(torch.sqrt(torch.clamp(eigvals, min=0)).to(torch.cfloat))

    # sqrt(rho) = V * sqrt(D) * V†
    sqrt_rho = eigvecs @ sqrt_D @ eigvecs.conj().T

    # Compute inner matrix
    inner = sqrt_rho @ sigma @ sqrt_rho

    # Eigenvalues of inner
    eigvals_inner, _ = torch.linalg.eigh(inner)
    fid = torch.sum(torch.sqrt(torch.clamp(eigvals_inner.real, min=0))) ** 2
    return fid.real
```

Step 8: Evaluate and Show Results

```python
import time

model.eval()
fid_list = []
td_list = []

with torch.no_grad():
    start_time = time.time()  # start timing for latency

    for i in range(len(X)):
        x_i = X[i].unsqueeze(0)  # batch size 1
        L_pred = model(x_i)
        rho_pred = reconstruct_density(L_pred)[0]  # get 2x2 matrix
        rho_true = Y[i]

        # Compute fidelity
        fid_val = fidelity(rho_pred, rho_true)
        fid_list.append(fid_val.item())

        # Compute trace distance
        td_val = trace_distance(rho_pred, rho_true)
        td_list.append(td_val.item())

    end_time = time.time()

# Metrics
mean_fid = np.mean(fid_list)
mean_td = np.mean(td_list)
latency_per_sample = (end_time - start_time) / len(X)

print(" Mean Fidelity:", mean_fid)
print(" Mean Trace Distance:", mean_td)
print("Average Inference Latency per sample (s):", latency_per_sample)
```

Visualize Example Density Matrices

```python
def plot_density_matrix(rho, title="Density Matrix"):
    """
    Plot the real and imaginary parts of a 2x2 density matrix.
    """
    rho = rho.numpy() if torch.is_tensor(rho) else rho
    fig, axes = plt.subplots(1, 2, figsize=(8, 4))

    axes[0].imshow(np.real(rho), cmap='viridis')
    axes[0].set_title(f"{title} (Real)")
    axes[0].set_xticks(range(rho.shape[0]))
    axes[0].set_yticks(range(rho.shape[1]))

    axes[1].imshow(np.imag(rho), cmap='viridis')
    axes[1].set_title(f"{title} (Imag)")
    axes[1].set_xticks(range(rho.shape[0]))
    axes[1].set_yticks(range(rho.shape[1]))

    plt.show()

# Show first 3 examples
with torch.no_grad():
    for i in range(3):
        L_pred = model(X[i].unsqueeze(0))
        rho_pred = reconstruct_density(L_pred)[0]
        rho_true = Y[i]

        print(f"Example {i+1}:")
        plot_density_matrix(rho_true, title="True ρ")
        plot_density_matrix(rho_pred, title="Predicted ρ")
```
